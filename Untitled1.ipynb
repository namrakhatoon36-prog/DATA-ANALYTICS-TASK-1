```python
import pandas as pd
import numpy as np
#1.Dataset load
url="https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df=pd.read_csv(url)

#original copy save (for recorde)
df_original=df.copy()

#2.inventory check
print("===INITIAL DATA STATUS===")
print("Missing value:\n",df.isnull().sum())
print("\nDuplicates:",df.duplicated().sum())
#3.Data cleaning operations
df['Age']=df['Age'].fillna(df['Age'].median())

#Embarked:mode
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
#cabin:Missing Flag('unknown')assign
df['Cabin']=df['Cabin'].fillna('Unknown')
#standardize text
df['Name']=df['Name'].str.strip()
df['Sex']=df['Sex'].str.lower().str.strip()

#4. Final Vrification check
print("\n===FINAL DATA STATUS===")
print("Remaining Missing values:\n",df.isnull().sum())

#5.cleaned file
df.to_csv('cleaned_titanic data set.csv',index=False)
```
